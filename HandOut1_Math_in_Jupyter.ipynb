# Writing Math in Jupyter Notebooks — A Quick Reference

In Jupyter, text cells use **Markdown** for formatting and **LaTeX** for equations. This guide covers everything you need for the problem sets.

To edit a text cell, **double-click** it. To render it, press **Shift + Enter**.

## 2 — Inline vs Display Math

LaTeX math is enclosed in dollar signs.

| Mode | Syntax | Use it for |
|---|---|---|
| **Inline** | `$ ... $` | Math within a sentence |
| **Display** | `$$ ... $$` | Equations on their own line, centred |

**Example:**

```
The normalisation condition is $\int |\psi|^2 dx = 1$, which means
the wavefunction must satisfy

$$\int_{-\infty}^{\infty} |\psi(x)|^2 \, dx = 1.$$
```

Renders as:

The normalisation condition is $\int |\psi|^2 dx = 1$, which means the wavefunction must satisfy

$$\int_{-\infty}^{\infty} |\psi(x)|^2 \, dx = 1.$$

## 5 — Integrals, Sums, and Derivatives

### Integrals

```latex
\int f(x)\, dx                          % indefinite
\int_{a}^{b} f(x)\, dx                  % definite
\int_{-\infty}^{\infty} |\psi|^2\, dx   % normalisation
```

$$\int_{-\infty}^{\infty} |\psi|^2\, dx$$

> **Tip:** `\,` adds a small space before $dx$ — standard mathematical style.

### Sums

```latex
\sum_{n=0}^{\infty} c_n \psi_n(x)
```

$$\sum_{n=0}^{\infty} c_n \psi_n(x)$$

### Derivatives

```latex
\frac{d\psi}{dx}            % first derivative
\frac{d^2\psi}{dx^2}        % second derivative
\frac{\partial \Psi}{\partial t}   % partial derivative
```

$$\frac{\partial \Psi}{\partial t}, \qquad \frac{d^2\psi}{dx^2}$$

## 6 — Quantum Mechanics Expressions

Here are complete examples of expressions that appear in Griffiths, ready to copy and modify.

### The time-dependent Schrödinger equation

```latex
i\hbar \frac{\partial \Psi}{\partial t} =
-\frac{\hbar^2}{2m}\frac{\partial^2 \Psi}{\partial x^2} + V\Psi
```

$$i\hbar \frac{\partial \Psi}{\partial t} = -\frac{\hbar^2}{2m}\frac{\partial^2 \Psi}{\partial x^2} + V\Psi$$

### Energy eigenvalue equation

```latex
\hat{H}\psi = E\psi
```

$$\hat{H}\psi = E\psi$$

### Expectation value of an operator

```latex
\langle \hat{Q} \rangle = \int_{-\infty}^{\infty} \Psi^* \hat{Q} \Psi \, dx
```

$$\langle \hat{Q} \rangle = \int_{-\infty}^{\infty} \Psi^* \hat{Q} \Psi \, dx$$

### Heisenberg uncertainty principle

```latex
\sigma_x \sigma_p \geq \frac{\hbar}{2}
```

$$\sigma_x \sigma_p \geq \frac{\hbar}{2}$$

### Commutator

```latex
[\hat{x}, \hat{p}] = i\hbar
```

$$[\hat{x}, \hat{p}] = i\hbar$$

### Gaussian wave packet

```latex
\psi(x) = \left(\frac{1}{2\pi\sigma^2}\right)^{1/4}
\exp\!\left(-\frac{(x-x_0)^2}{4\sigma^2}\right) e^{ik_0 x}
```

$$\psi(x) = \left(\frac{1}{2\pi\sigma^2}\right)^{1/4} \exp\!\left(-\frac{(x-x_0)^2}{4\sigma^2}\right) e^{ik_0 x}$$

### Bra-ket notation

```latex
\langle \psi | \phi \rangle      % inner product
| n \rangle                      % ket
\langle n |                      % bra
\langle m | \hat{H} | n \rangle  % matrix element
```

$$\langle \psi | \phi \rangle, \quad | n \rangle, \quad \langle m | \hat{H} | n \rangle$$

## 7 — Aligning Multi-Line Equations

Use the `align` environment when showing steps in a derivation. Use `&` to set the alignment point and `\\` to end each line.

```latex
\begin{align}
\langle x^2 \rangle &= \int_{-\infty}^{\infty} x^2 |\psi|^2 \, dx \\
                    &= \frac{\hbar}{2m\omega}\left(2n + 1\right) \\
                    &= \frac{\hbar}{m\omega}\left(n + \frac{1}{2}\right)
\end{align}
```

$$\begin{align}
\langle x^2 \rangle &= \int_{-\infty}^{\infty} x^2 |\psi|^2 \, dx \\
                    &= \frac{\hbar}{2m\omega}\left(2n + 1\right) \\
                    &= \frac{\hbar}{m\omega}\left(n + \frac{1}{2}\right)
\end{align}$$

## 9 — Common Mistakes

| Mistake | Wrong | Right |
|---|---|---|
| Multi-character exponent without braces | `e^ikx` → $e^ikx$ | `e^{ikx}` → $e^{ikx}$ |
| Forgetting `\` before function names | `sin(x)` → $sin(x)$ | `\sin(x)` → $\sin(x)$ |
| Using `*` for conjugate | `\psi*` → $\psi*$ | `\psi^*` → $\psi^*$ |
| Absolute value bars | `|\psi|` → $|\psi|$ | `\lvert\psi\rvert` → $\lvert\psi\rvert$ (or `|\psi|` works in most renderers) |
| Tall brackets not scaling | `(\frac{a}{b})` → $(\frac{a}{b})$ | `\left(\frac{a}{b}\right)` → $\left(\frac{a}{b}\right)$ |
