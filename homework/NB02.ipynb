````python
# ==============================================================
# PHYS 3316 — Quantum Mechanics I
#
# Notebook 1 : Infinite Square Well
# Computational Exercise:
#     Time Evolution of an Arbitrary Superposition
#
# Author:
# Date:
#
# Description
# -----------
# This program constructs and visualizes the time-dependent
# wavefunction of a particle confined to a one-dimensional
# infinite square well.  The wavefunction is represented as a
# linear combination of stationary energy eigenstates,
#
#        Ψ(x,t) = Σ cₙ ψₙ(x) exp(-iEₙt),
#
# where the expansion coefficients cₙ may be chosen by the user.
#
# The program computes
#
#   • stationary eigenfunctions ψₙ(x)
#   • energy eigenvalues Eₙ
#   • time-dependent wavefunction Ψ(x,t)
#   • probability density |Ψ(x,t)|²
#   • expectation values <x>(t) and <p>(t)
#   • phase-space trajectory (<x>,<p>)
#
# All calculations are performed in atomic units.
#
# Atomic Units
# ------------
#      ħ  = 1
#      mₑ = 1
#      e  = 1
#
# Length  : Bohr radius (a₀)
# Energy  : Hartree (E_h)
# Time    : ħ/E_h
#
# Students are encouraged to modify:
#
#   • box length
#   • expansion coefficients
#   • number of basis states
#   • plotting parameters
#
# and investigate how these changes affect the quantum dynamics.
#
# ==============================================================


import numpy as np
import matplotlib.pyplot as plt

# ==============================================================

# USER PARAMETERS

# ==============================================================

# Infinite well width (Bohr radii)

L = 20.0

# Expansion coefficients
#
# Psi(x,0) = Σ c_n ψ_n(x)
#
# Example:
#
#    c = [1,1]          equal superposition (example)
#    c = [1,0]          ground state
#    c = [0,1]          first excited state
#    c = [1,1j]         relative phase π/2
#

coefficients = np.array([
    1.0,
    1.0
], dtype=complex)

# Number of spatial grid points
Nx = 1000
# Number of time points
Nt = 500

# Example with relative phase:
#
# coefficients = np.array([
#     1.0,
#     1.0j
# ], dtype=complex)

# Times at which the wavefunction will be plotted.
# These are entered below after a reference period is calculated.


# ============================================================
# ATOMIC-UNIT CONVERSIONS
# ============================================================

hartree_to_ev = 27.211386
atomic_time_to_fs = 2.418884e-2
bohr_to_nm = 5.291772e-2


# ============================================================
# INFINITE-SQUARE-WELL FUNCTIONS
# ============================================================

def psi_n(n, x, L):
    """
    Normalized eigenfunction of the infinite square well.

    Parameters
    ----------
    n : int
        Quantum number, n = 1, 2, 3, ...
    x : ndarray
        Position in Bohr radii.
    L : float
        Box length in Bohr radii.

    Returns
    -------
    ndarray
        Values of psi_n(x).
    """
    return np.sqrt(2.0 / L) * np.sin(n * np.pi * x / L)


def energy_n(n, L):
    """
    Infinite-square-well energy in Hartrees.
    """
    return n**2 * np.pi**2 / (2.0 * L**2)


def normalize_coefficients(coefficients):
    """
    Normalize the expansion coefficients so that

        sum_n |c_n|^2 = 1.
    """
    norm = np.sqrt(np.sum(np.abs(coefficients)**2))

    if norm == 0:
        raise ValueError("At least one expansion coefficient must be nonzero.")

    return coefficients / norm


# Normalize the user-supplied coefficients
coefficients = normalize_coefficients(coefficients)

# Number of states included in the expansion
number_of_states = len(coefficients)

# Quantum numbers n = 1, 2, ..., number_of_states
quantum_numbers = np.arange(1, number_of_states + 1)

# Corresponding energies
energies = np.array([
    energy_n(n, L) for n in quantum_numbers
])


def Psi(x, t, coefficients, energies, L):
    """
    Construct the time-dependent wavefunction

        Psi(x,t) = sum_n c_n psi_n(x) exp(-i E_n t).

    Atomic units are used, so hbar = 1.
    """
    wavefunction = np.zeros_like(x, dtype=complex)

    for n, c_n, E_n in zip(
        quantum_numbers,
        coefficients,
        energies
    ):
        wavefunction += (
            c_n
            * psi_n(n, x, L)
            * np.exp(-1j * E_n * t)
        )

    return wavefunction



# ============================================================
# EXPECTATION VALUES AS FUNCTIONS OF TIME
# ============================================================

# Number of time points used for the calculation
number_of_time_points = Nt

# Follow the system over one reference period
time_grid = np.linspace(
    0.0,
    reference_period,
    number_of_time_points
)


def expectation_x(x, wavefunction):
    """
    Calculate the expectation value <x>.

    Parameters
    ----------
    x : ndarray
        Spatial grid in Bohr radii.
    wavefunction : ndarray
        Psi(x,t) evaluated on the spatial grid.

    Returns
    -------
    float
        Expectation value <x> in Bohr radii.
    """
    probability_density = np.abs(wavefunction)**2

    return np.trapezoid(
        x * probability_density,
        x
    )


def expectation_p(x, wavefunction):
    """
    Calculate the expectation value <p> using

        p = -i d/dx

    in atomic units.
    """
    dpsi_dx = np.gradient(
        wavefunction,
        x,
        edge_order=2
    )

    integrand = (
        np.conjugate(wavefunction)
        * (-1j * dpsi_dx)
    )

    p_average = np.trapezoid(
        integrand,
        x
    )

    # The exact expectation value is real because momentum is
    # a Hermitian operator. A small imaginary component may remain
    # because of numerical differentiation and integration.
    return np.real(p_average)


# Arrays that will hold the expectation values
x_average = np.zeros_like(time_grid)
p_average = np.zeros_like(time_grid)

for index, t in enumerate(time_grid):

    wavefunction = Psi(
        x,
        t,
        coefficients,
        energies,
        L
    )

    x_average[index] = expectation_x(
        x,
        wavefunction
    )

    p_average[index] = expectation_p(
        x,
        wavefunction
    )


# ============================================================
# DISPLAY THE STATE
# ============================================================

print(f"Box length: {L:.3f} a0")
print(f"Box length: {L * bohr_to_nm:.3f} nm")
print()

print("Normalized expansion coefficients")
print("---------------------------------")

for n, c_n in zip(quantum_numbers, coefficients):
    probability = np.abs(c_n)**2

    print(
        f"n = {n:2d}: "
        f"c_n = {c_n.real:+.4f}{c_n.imag:+.4f}j, "
        f"|c_n|^2 = {probability:.4f}"
    )

print()
print(f"Sum of probabilities = {np.sum(np.abs(coefficients)**2):.6f}")

print()
print("Energy levels")
print("-------------")

for n, E_n in zip(quantum_numbers, energies):
    print(
        f"E_{n} = {E_n:.6f} Hartree "
        f"= {E_n * hartree_to_ev:.4f} eV"
    )


# ============================================================
# CHOOSE A REFERENCE TIME SCALE
# ============================================================

# Identify the populated states
tolerance = 1.0e-12
populated = np.where(np.abs(coefficients) > tolerance)[0]

if len(populated) >= 2:
    populated_energies = energies[populated]

    # All positive energy differences among populated states
    energy_gaps = []

    for i in range(len(populated_energies)):
        for j in range(i + 1, len(populated_energies)):
            gap = abs(
                populated_energies[j]
                - populated_energies[i]
            )

            if gap > tolerance:
                energy_gaps.append(gap)

    smallest_gap = min(energy_gaps)

    reference_period = 2.0 * np.pi / smallest_gap

    print()
    print(
        "Reference period based on the smallest populated "
        "energy gap:"
    )
    print(
        f"T_ref = {reference_period:.6f} atomic units"
    )
    print(
        f"T_ref = "
        f"{reference_period * atomic_time_to_fs:.6f} fs"
    )

else:
    reference_period = 1.0

    print()
    print(
        "Only one energy eigenstate is populated, so the "
        "probability density is stationary."
    )


# ============================================================
# TIMES TO PLOT
# ============================================================

times = np.array([
    0.0,
    reference_period / 4.0,
    reference_period / 2.0,
    3.0 * reference_period / 4.0,
    reference_period
])

time_labels = [
    r"$t=0$",
    r"$t=T_{\rm ref}/4$",
    r"$t=T_{\rm ref}/2$",
    r"$t=3T_{\rm ref}/4$",
    r"$t=T_{\rm ref}$"
]


# ============================================================
# SPATIAL GRID
# ============================================================

x = np.linspace(0.0, L,Nx)


# ============================================================
# PLOT THE INITIAL PROBABILITY DENSITY
# ============================================================

initial_wavefunction = Psi(
    x,
    0.0,
    coefficients,
    energies,
    L
)

plt.figure(figsize=(8, 5))

plt.plot(
    x / L,
    np.abs(initial_wavefunction)**2
)

plt.xlabel(r"$x/L$")
plt.ylabel(r"$|\Psi(x,0)|^2$")
plt.title("Initial Probability Density")
plt.tight_layout()
plt.show()


# ============================================================
# PLOT THE REAL PART
# ============================================================

plt.figure(figsize=(8, 5))

for t, label in zip(times, time_labels):
    wavefunction = Psi(
        x,
        t,
        coefficients,
        energies,
        L
    )

    plt.plot(
        x / L,
        np.real(wavefunction),
        label=label
    )

plt.axhline(0.0, linewidth=0.8)
plt.xlabel(r"$x/L$")
plt.ylabel(r"$\mathrm{Re}\,\Psi(x,t)$")
plt.title("Real Part of the Wavefunction")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# PLOT THE IMAGINARY PART
# ============================================================

plt.figure(figsize=(8, 5))

for t, label in zip(times, time_labels):
    wavefunction = Psi(
        x,
        t,
        coefficients,
        energies,
        L
    )

    plt.plot(
        x / L,
        np.imag(wavefunction),
        label=label
    )

plt.axhline(0.0, linewidth=0.8)
plt.xlabel(r"$x/L$")
plt.ylabel(r"$\mathrm{Im}\,\Psi(x,t)$")
plt.title("Imaginary Part of the Wavefunction")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# PLOT THE PROBABILITY DENSITY
# ============================================================

plt.figure(figsize=(8, 5))

for t, label in zip(times, time_labels):
    wavefunction = Psi(
        x,
        t,
        coefficients,
        energies,
        L
    )

    probability_density = np.abs(wavefunction)**2

    plt.plot(
        x / L,
        probability_density,
        label=label
    )

plt.xlabel(r"$x/L$")
plt.ylabel(r"$|\Psi(x,t)|^2$")
plt.title("Time Evolution of the Probability Density")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# CHECK NORMALIZATION
# ============================================================

print()
print("Numerical normalization check")
print("-----------------------------")

for t, label in zip(times, time_labels):
    wavefunction = Psi(
        x,
        t,
        coefficients,
        energies,
        L
    )

    normalization = np.trapezoid(
        np.abs(wavefunction)**2,
        x
    )

    print(
        f"t = {t:10.4f} a.u.: "
        f"integral |Psi|^2 dx = {normalization:.8f}"
    )



````

## Problem 6 — Expectation Values and Ehrenfest’s Theorem

In the previous problem, you constructed a time-dependent wavefunction of the form

$$
\Psi(x,t)
=
\sum_{n=1}^{N}
c_n\psi_n(x)e^{-iE_nt}.
$$

Although the probability density changes with time, the motion is not described by a definite particle trajectory. Instead, we can follow the time evolution of expectation values.

The expectation value of position is

$$
\langle x\rangle(t)
=
\int_0^L
\Psi^*(x,t)\,x\,\Psi(x,t)\,dx.
$$

The expectation value of momentum is

$$
\langle p\rangle(t)
=
\int_0^L
\Psi^*(x,t)
\left(
-i\frac{\partial}{\partial x}
\right)
\Psi(x,t)\,dx,
$$

where atomic units have been used so that $\hbar=1$.

$$

$$

$$

$$
