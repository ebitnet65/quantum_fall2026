```python
import numpy as np
import matplotlib.pyplot as plt

from math import factorial
from scipy.integrate import simpson
from scipy.signal import find_peaks
from scipy.special import eval_genlaguerre

np.set_printoptions(precision=6, suppress=True)

# Atomic units are used in the computational sections:
# a0 = hbar = m_e = e = 1.
a0 = 1.0


def radial_hydrogen(n, ell, r, Z=1.0, a0=1.0):
    """Normalized hydrogenic radial wavefunction R_{n,ell}(r).

    Parameters
    ----------
    n : int
        Principal quantum number, n >= 1.
    ell : int
        Orbital angular-momentum quantum number, 0 <= ell < n.
    r : array_like
        Radius in the same length units as a0.
    Z : float
        Nuclear charge number.
    a0 : float
        Bohr radius.

    Returns
    -------
    ndarray
        R_{n,ell}(r), normalized so that integral r^2 |R|^2 dr = 1.
    """
    if n < 1 or ell < 0 or ell >= n:
        raise ValueError("Require n >= 1 and 0 <= ell < n.")

    r = np.asarray(r, dtype=float)
    rho = 2.0 * Z * r / (n * a0)

    normalization = (
        (2.0 * Z / (n * a0)) ** 1.5
        * np.sqrt(
            factorial(n - ell - 1)
            / (2.0 * n * factorial(n + ell))
        )
    )

    laguerre = eval_genlaguerre(
        n - ell - 1,
        2 * ell + 1,
        rho,
    )

    return (
        normalization
        * np.exp(-rho / 2.0)
        * rho**ell
        * laguerre
    )


def radial_probability(n, ell, r, Z=1.0, a0=1.0):
    """Radial probability density P_{n,ell}(r)."""
    R = radial_hydrogen(n, ell, r, Z=Z, a0=a0)
    return r**2 * np.abs(R)**2


# A grid large enough for the n <= 3 hydrogen states used below.
r = np.linspace(0.0, 40.0 * a0, 6001)

states = [
    (1, 0, "1s"),
    (2, 0, "2s"),
    (2, 1, "2p"),
    (3, 0, "3s"),
    (3, 1, "3p"),
    (3, 2, "3d"),
]
```
