## Problem 4 — Ehrenfest's Theorem for the Harmonic Oscillator

For a particle moving in one dimension, the expectation value of the momentum is

$$
\langle p\rangle
=
\int_{-\infty}^{\infty}
\Psi^*(x,t)
\left(
-i\hbar\frac{\partial}{\partial x}
\right)
\Psi(x,t)\,dx.
$$

The time-dependent Schrödinger equation is

$$
i\hbar\frac{\partial \Psi}{\partial t}
=
-\frac{\hbar^2}{2m}
\frac{\partial^2\Psi}{\partial x^2}
+
V(x)\Psi.
$$

Assume that $\Psi(x,t)$ and its spatial derivatives vanish sufficiently rapidly as $x\to\pm\infty$.

### (a) Time derivative of the momentum expectation value

Differentiate the momentum expectation value with respect to time:

$$
\frac{d\langle p\rangle}{dt}
=
\frac{d}{dt}
\int_{-\infty}^{\infty}
\Psi^*
\left(
-i\hbar\frac{\partial}{\partial x}
\right)
\Psi\,dx.
$$

Use the product rule to show that

$$
\frac{d\langle p\rangle}{dt}
=
\int_{-\infty}^{\infty}
\left[
\frac{\partial\Psi^*}{\partial t}
\left(
-i\hbar\frac{\partial\Psi}{\partial x}
\right)
+
\Psi^*
\left(
-i\hbar\frac{\partial}{\partial x}
\right)
\frac{\partial\Psi}{\partial t}
\right]dx.
$$

### (b) Operator form of the equation of motion

The expression obtained above may be written in Dirac notation as
$$
\frac{d\langle p\rangle}{dt}
=
\left\langle
\frac{\partial\Psi}{\partial t}
\middle|
\hat p
\middle|
\Psi
\right\rangle
+
\left\langle
\Psi
\middle|
\hat p
\middle|
\frac{\partial\Psi}{\partial t}
\right\rangle,
$$
where
$$
\hat p=-i\hbar\frac{\partial}{\partial x}.
$$

The time-dependent Schrödinger equation and its adjoint may be written as
$$
i\hbar\frac{\partial}{\partial t}|\Psi\rangle
=
\hat H|\Psi\rangle
$$
and
$$
-i\hbar\frac{\partial}{\partial t}\langle\Psi|
=
\langle\Psi|\hat H.
$$
Use these equations to show that
$$
\frac{\partial}{\partial t}|\Psi\rangle
=
-\frac{i}{\hbar}\hat H|\Psi\rangle
$$
and
$$
\frac{\partial}{\partial t}\langle\Psi|
=
\frac{i}{\hbar}\langle\Psi|\hat H.
$$
Substitute these results into the time derivative of the momentum expectation value and show that
$$
\frac{d\langle p\rangle}{dt}
=
\frac{i}{\hbar}
\langle\Psi|\hat H\hat p|\Psi\rangle
-
\frac{i}{\hbar}
\langle\Psi|\hat p\hat H|\Psi\rangle.
$$
Using the commutator definition

$$
[\hat H,\hat p]
=
\hat H\hat p-\hat p\hat H,
$$

deduce that

$$
\boxed{
\frac{d\langle p\rangle}{dt}
=
\frac{i}{\hbar}
\langle\Psi|[\hat H,\hat p]|\Psi\rangle
}
$$

or, equivalently,

$$
\boxed{
\frac{d\langle p\rangle}{dt}
=
\frac{1}{i\hbar}
\langle\Psi|[\hat p,\hat H]|\Psi\rangle.
}
$$

This is the Heisenberg equation of motion for the expectation value of an operator with no explicit time dependence.

For a general operator $\hat A(t)$ that may depend explicitly on time, the corresponding equation is

$$
\boxed{
\frac{d\langle A\rangle}{dt}
=
\frac{i}{\hbar}
\langle[\hat H,\hat A]\rangle
+
\left\langle
\frac{\partial\hat A}{\partial t}
\right\rangle.
}
$$
Explain why the final term vanishes for the momentum operator $\hat p$ used here.
