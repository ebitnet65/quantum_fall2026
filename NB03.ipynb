## Overview

The quantum harmonic oscillator is one of the most important exactly solvable models in quantum mechanics. It provides a useful approximation for systems undergoing small oscillations near a stable equilibrium and appears throughout atomic, molecular, optical, and condensed-matter physics.

The Hamiltonian is

$$
\hat{H}
=
\frac{\hat{p}^2}{2m}
+
\frac{1}{2}m\omega^2\hat{x}^2.
$$

Rather than solving the Schrödinger equation directly as a differential equation, we will develop the problem primarily using the **raising and lowering operator method**.

The harmonic oscillator also provides a natural introduction to the **variational theorem**. A Gaussian trial wavefunction can be optimized to minimize the expectation value of the energy. For the quadratic potential, this procedure reproduces the exact ground state.

## Physical Model

The harmonic-oscillator potential is

$$
V(x)
=
\frac{1}{2}m\omega^2x^2,
$$

where

- $m$ is the particle mass,
- $\omega$ is the angular frequency, and
- $x=0$ is the equilibrium position.

The potential is symmetric about the origin and increases quadratically as the particle moves away from equilibrium.

Classically, the particle oscillates between the turning points

$$
x_{\mathrm{tp}}
=
\pm
\sqrt{\frac{2E}{m\omega^2}}.
$$

Quantum mechanically, the allowed energies are discrete:

$$
E_n
=
\hbar\omega
\left(
n+\frac{1}{2}
\right),
\qquad
n=0,1,2,\ldots
$$

The lowest possible energy is therefore

$$
E_0
=
\frac{1}{2}\hbar\omega.
$$

This nonzero ground-state energy is called the **zero-point energy**.

---

## Natural Harmonic-Oscillator Units

The harmonic oscillator defines natural scales for position, momentum, energy, and time.

The characteristic length is

$$
x_0
=
\sqrt{\frac{\hbar}{m\omega}}.
$$

The characteristic momentum is

$$
p_0
=
\sqrt{m\hbar\omega}.
$$

The natural energy scale is

$$
E_0
=
\hbar\omega,
$$

and the natural time scale is

$$
t_0
=
\frac{1}{\omega}.
$$

It is often convenient to introduce the dimensionless coordinate

$$
\xi
=
\frac{x}{x_0}
=
\sqrt{\frac{m\omega}{\hbar}}\,x.
$$

In terms of $\xi$, the Hamiltonian becomes

$$
\frac{\hat{H}}{\hbar\omega}
=
-\frac{1}{2}
\frac{d^2}{d\xi^2}
+
\frac{1}{2}\xi^2.
$$

Unless otherwise stated, the computational sections of this notebook use dimensionless harmonic-oscillator units:

$$
\hbar=1,
\qquad
m=1,
\qquad
\omega=1.
$$

With this convention,

$$
\hat{H}
=
-\frac{1}{2}\frac{d^2}{dx^2}
+
\frac{1}{2}x^2
$$

and

$$
E_n
=
n+\frac{1}{2}.
$$

## Raising and Lowering Operators

Define the lowering operator

$$
\hat{a}
=
\sqrt{\frac{m\omega}{2\hbar}}\,\hat{x}
+
\frac{i}{\sqrt{2m\hbar\omega}}\,\hat{p},
$$

and the raising operator

$$
\hat{a}^{\dagger}
=
\sqrt{\frac{m\omega}{2\hbar}}\,\hat{x}
-
\frac{i}{\sqrt{2m\hbar\omega}}\,\hat{p}.
$$

> Note that the text book uses a rather unstandard notation of $a_-$ for the lowering operator and $a_+$ as the raising operator.  In my lectures and notes, I'll use the more commonly adapted notation used here. 


These operators satisfy the commutation relation

$$
[\hat{a},\hat{a}^{\dagger}]
=
1.
$$

The Hamiltonian can then be written as

$$
\hat{H}
=
\hbar\omega
\left(
\hat{a}^{\dagger}\hat{a}
+
\frac{1}{2}
\right).
$$

The operators act on the energy eigenstates according to

$$
\hat{a}|n\rangle
=
\sqrt{n}\,|n-1\rangle,
$$

and

$$
\hat{a}^{\dagger}|n\rangle
=
\sqrt{n+1}\,|n+1\rangle.
$$

The position and momentum operators can be expressed as

$$
\hat{x}
=
\sqrt{\frac{\hbar}{2m\omega}}
\left(
\hat{a}
+
\hat{a}^{\dagger}
\right),
$$

and

$$
\hat{p}
=
-i
\sqrt{\frac{m\hbar\omega}{2}}
\left(
\hat{a}
-
\hat{a}^{\dagger}
\right).
$$

These relations make it possible to calculate many expectation values and matrix elements without directly evaluating integrals.

## Stationary-State Wavefunctions

The normalized position-space eigenfunctions are

$$
\psi_n(x)
=
\frac{1}{\sqrt{2^n n!}}
\left(
\frac{m\omega}{\pi\hbar}
\right)^{1/4}
H_n(\xi)
e^{-\xi^2/2},
$$

where

$$
\xi
=
\sqrt{\frac{m\omega}{\hbar}}\,x
$$

and $H_n(\xi)$ is the Hermite polynomial of degree $n$.

The first few Hermite polynomials are

$$
H_0(\xi)=1,
$$

$$
H_1(\xi)=2\xi,
$$

$$
H_2(\xi)=4\xi^2-2,
$$

$$
H_3(\xi)=8\xi^3-12\xi.
$$

The parity of the eigenfunctions alternates:

$$
\psi_n(-x)
=
(-1)^n\psi_n(x).
$$

Therefore,

- even $n$ states have even parity,
- odd $n$ states have odd parity.

The state $\psi_n$ has exactly $n$ nodes.

---

## Variational Principle

The variational theorem states that for any normalized trial wavefunction $\phi$,

$$
E[\phi]
=
\langle\phi|\hat{H}|\phi\rangle
\ge E_0,
$$

where $E_0$ is the exact ground-state energy.

A trial wavefunction may contain one or more adjustable parameters. The best approximation within the chosen family of trial functions is obtained by minimizing the energy with respect to those parameters.

For the harmonic oscillator, we will use a normalized Gaussian trial function,

$$
\phi_\alpha(x)
=
\left(
\frac{2\alpha}{\pi}
\right)^{1/4}
e^{-\alpha x^2},
\qquad
\alpha>0.
$$

The variational energy is

$$
E(\alpha)
=
\langle\phi_\alpha|\hat{H}|\phi_\alpha\rangle.
$$

The parameter $\alpha$ controls the width of the Gaussian:

- large $\alpha$ produces a narrow wavefunction,
- small $\alpha$ produces a broad wavefunction.

The optimum width results from a balance between kinetic and potential energy.

```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.integrate import simpson
from scipy.optimize import minimize_scalar
from scipy.special import eval_hermite, factorial

# Problem 1 — Raising and Lowering Operators

The lowering and raising operators for the one-dimensional harmonic oscillator are

$$
\hat{a}
=
\sqrt{\frac{m\omega}{2\hbar}}\hat{x}
+
\frac{i}{\sqrt{2m\hbar\omega}}\hat{p},
$$

and

$$
\hat{a}^{\dagger}
=
\sqrt{\frac{m\omega}{2\hbar}}\hat{x}
-
\frac{i}{\sqrt{2m\hbar\omega}}\hat{p}.
$$

The Hamiltonian is

$$
\hat{H}
=
\frac{\hat{p}^2}{2m}
+
\frac12m\omega^2\hat{x}^2.
$$

### (a)

Show that

$$
[\hat{a},\hat{a}^{\dagger}]=1.
$$

---

### (b)

Express the Hamiltonian in terms of the ladder operators and show that

$$
\hat{H}
=
\hbar\omega
\left(
\hat{a}^{\dagger}\hat{a}
+\frac12
\right).
$$

---

### (c)

Suppose

$$
\hat{a}|0\rangle=0.
$$

Determine the ground-state energy.

---

### (d)

Show that repeated application of the raising operator generates an infinite sequence of normalized eigenstates,

$$
|n\rangle
=
\frac{(\hat a^\dagger)^n}{\sqrt{n!}}
|0\rangle.
$$

Hence determine the complete energy spectrum.

---

### (e)

Explain why the harmonic oscillator possesses a nonzero ground-state energy. Relate your answer to the uncertainty principle.

# SOLUTION:

----

# Problem 2 — Matrix Elements and Selection Rules

Using the ladder operators, show that

$$
\hat{x}
=
\sqrt{\frac{\hbar}{2m\omega}}
(\hat a+\hat a^\dagger)
$$

and

$$
\hat{p}
=
-i
\sqrt{\frac{m\hbar\omega}{2}}
(\hat a-\hat a^\dagger).
$$

### (a)

Calculate the matrix elements

$$
\langle m|\hat{x}|n\rangle.
$$

---

### (b)

Calculate

$$
\langle m|\hat{p}|n\rangle.
$$

---

### (c)

Construct the matrices representing $\hat{x}$ and $\hat{p}$ using the basis

$$
|0\rangle,
|1\rangle,
|2\rangle,
|3\rangle,
|4\rangle.
$$

---

### (d)

Verify explicitly that

$$
[\hat{x},\hat{p}]
=
i\hbar I
$$

within this truncated basis.

How does the accuracy of the commutator change as the basis size is increased?

---

### (e)

From your results, explain why electric-dipole transitions in the harmonic oscillator satisfy the selection rule

$$
\Delta n=\pm1.
$$

SOLUTION:

---

# Problem 3 — The Harmonic Oscillator Wavefunctions

Although the energy spectrum of the harmonic oscillator can be obtained entirely using the ladder-operator formalism, it is useful to examine the corresponding wavefunctions in position space.

The normalized stationary-state wavefunctions are
$$
\psi_n(x)
=
\frac{1}{\sqrt{2^n n!}}
\left(
\frac{m\omega}{\pi\hbar}
\right)^{1/4}
H_n(\xi)
e^{-\xi^2/2},
$$
where
$$
\xi
=
\sqrt{\frac{m\omega}{\hbar}}\,x
$$
and $H_n(\xi)$ is the Hermite polynomial of degree $n$.

Unless otherwise stated, use dimensionless harmonic-oscillator units,
$$
\hbar = m = \omega = 1.
$$

---

## Physical Objective

Construct and visualize the stationary-state wavefunctions of the quantum harmonic oscillator. Examine how their shape, parity, and spatial extent change with increasing energy.

---

## Mathematical Background

The energy eigenvalues are
$$
E_n
=
n+\frac12,
\qquad
n=0,1,2,\ldots
$$
in dimensionless units.

The classical turning points satisfy
$$
\frac12x_{\rm tp}^2
=
E_n,
$$
or
$$
x_{\rm tp}
=
\pm\sqrt{2E_n}.
$$

As the energy increases, the wavefunctions become more oscillatory and extend farther from the origin.

---

## Computational Tasks

### (a) Plot the first six stationary states

Plot
$$
\psi_n(x),
\qquad
n=0,1,\ldots,5,
$$
on the same figure.
---

### (b) Plot the corresponding probability densities

Plot
$$
|\psi_n(x)|^2
$$
for the same six states.
What differences do you observe between the wavefunctions and the probability densities?

---

### (c) Energy-level diagram

Construct a figure showing
- the harmonic oscillator potential,
$$
V(x)=\frac12x^2,
$$

- the first six energy levels,
$$
E_n=n+\frac12,
$$

- and the corresponding wavefunctions drawn at their respective energies.
This is one of the standard visualizations of the harmonic oscillator.

---

### (d) Classical turning points

For each state, compute the classical turning points
$$
x_{\rm tp}
=
\pm\sqrt{2E_n}.
$$
Indicate these turning points on your probability-density plots.

How much of the quantum probability density lies outside the classically allowed region?

---

### (e) Nodes and parity

Determine the number of nodes for each wavefunction.

Verify that
- even values of $n$ produce even wavefunctions,
$$
\psi_n(-x)=\psi_n(x),
$$
- odd values of $n$ produce odd wavefunctions,
$$
\psi_n(-x)=-\psi_n(x).
$$

---

## Expected Output

Your notebook should include
- plots of the first six wavefunctions,
- plots of the first six probability densities,
- an energy-level diagram with wavefunctions,
- classical turning points for each state.

---

## Physical Interpretation

Answer the following questions.
1. How does the number of nodes depend on the quantum number?
2. Why does the spatial extent of the wavefunction increase with increasing energy?
3. Why is there a finite probability of finding the particle outside the classical turning points?
4. How does the harmonic oscillator differ from the infinite square well?
5. Which features of the wavefunctions become increasingly "classical" as $n$ becomes large?

---

## ★ Challenge

The Hermite polynomials satisfy the recursion relation

$$
H_{n+1}(x)
=
2xH_n(x)-2nH_{n-1}(x).
$$

Write your own Python routine that generates the Hermite polynomials recursively instead of using `scipy.special.eval_hermite`.

Compare your results with SciPy for the first ten wavefunctions.

# SOLUTION:


---

# Problem 4 — Orthogonality and Basis Convergence

The harmonic-oscillator eigenfunctions form an orthonormal set:

$$
\int_{-\infty}^{\infty}
\psi_m^*(x)\psi_n(x)\,dx
=
\delta_{mn}.
$$

In numerical calculations, however, the integral must be evaluated over a finite spatial interval using a finite grid. The result therefore depends on both the size of the computational domain and the grid resolution.

---

## Physical Objective

Verify the normalization and orthogonality of the harmonic-oscillator eigenfunctions numerically. Examine how the accuracy of the calculation depends on the spatial range and grid spacing.

---

## Mathematical Background

Define the overlap matrix

$$
S_{mn}
=
\int_{-\infty}^{\infty}
\psi_m^*(x)\psi_n(x)\,dx.
$$

For an exact orthonormal basis,

$$
S_{mn}
=
\delta_{mn}.
$$

Thus, the exact overlap matrix is the identity matrix:

$$
S=I.
$$

In a numerical calculation, the integral is approximated over a finite interval,

$$
-x_{\max}\le x\le x_{\max},
$$

and evaluated on a grid containing $N_x$ points.

The numerical overlap matrix will therefore differ slightly from the identity.

---

## Computational Tasks

### (a) Verify normalization

For the first six harmonic-oscillator states, calculate

$$
\int_{-x_{\max}}^{x_{\max}}
|\psi_n(x)|^2\,dx.
$$

Use an appropriate numerical integration routine, such as `scipy.integrate.simpson`.

Report the normalization integral for each state.

---

### (b) Construct the overlap matrix

Calculate

$$
S_{mn}
=
\int_{-x_{\max}}^{x_{\max}}
\psi_m^*(x)\psi_n(x)\,dx
$$

for

$$
m,n=0,1,\ldots,5.
$$

Display the result as a numerical matrix.

The matrix should be approximately equal to the identity matrix.

---

### (c) Measure the numerical error

Define the error matrix

$$
\Delta S
=
S-I.
$$

Calculate the maximum absolute error,

$$
\epsilon_{\max}
=
\max_{m,n}
\left|
S_{mn}-\delta_{mn}
\right|.
$$

Also calculate the Frobenius norm,

$$
\epsilon_F
=
\sqrt{
\sum_{m,n}
\left|
S_{mn}-\delta_{mn}
\right|^2
}.
$$

Report both errors.

---

### (d) Convergence with spatial range

Hold the number of grid points fixed and repeat the calculation for several values of $x_{\max}$.

For example,

```python
x_max_values = [3.0, 4.0, 5.0, 6.0, 8.0]
```

## Basis Completeness

Orthogonality and completeness are related but distinct properties.

Orthogonality means that

$$
\langle m|n\rangle
=
\delta_{mn}.
$$

Completeness means that an arbitrary physical state can be represented as

$$
|\Psi\rangle
=
\sum_{n=0}^{\infty}
c_n|n\rangle.
$$

A finite numerical basis contains only the first $N$ eigenstates:

$$
|\Psi\rangle
\approx
\sum_{n=0}^{N-1}
c_n|n\rangle.
$$

The accuracy of this approximation must be tested by increasing $N$ until the quantities of interest no longer change significantly.
