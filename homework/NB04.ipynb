## Overview

The infinite square well introduced the idea that boundary conditions lead naturally to quantized energy levels. In this notebook, we consider two more realistic one-dimensional potentials: the attractive delta-function potential and the finite square well.

Unlike the infinite well, these systems allow the wavefunction to extend into classically forbidden regions, introducing the important quantum phenomenon of **tunneling**. Although both models are idealized, they capture many of the essential features of bound states in atoms, molecules, semiconductor quantum wells, and nanoscale devices.

The attractive delta-function potential provides the simplest example of a localized potential that supports a bound state. Despite its simplicity, it illustrates how an arbitrarily weak attractive potential can bind a particle in one dimension.

The finite square well extends these ideas by introducing a potential of finite width and depth. In this case, the Schrödinger equation must be solved in multiple regions and the solutions matched at the boundaries. The resulting transcendental equations generally require numerical methods to determine the allowed bound-state energies.

Throughout this notebook, analytical derivations are combined with computational methods to investigate the properties of bound states and tunneling.

## Physical Models

### Attractive Delta-Function Potential

The attractive delta-function potential is

$$
V(x)=-\alpha\,\delta(x),
\qquad \alpha>0,
$$

where $\delta(x)$ is the Dirac delta function.

Although infinitely narrow, the potential supports exactly one bound state.

---

### Finite Square Well

The finite square well is defined by

$$
V(x)=
\begin{cases}
-V_0, & |x|<a, \\
0, & |x|\ge a,
\end{cases}
$$

where

- $V_0$ is the well depth,
- $2a$ is the well width.

Unlike the infinite well, the wavefunction extends beyond the boundaries of the well and decays exponentially outside the classically allowed region.

---

## Key Concepts

This notebook introduces several fundamental ideas that appear throughout quantum mechanics.

### Localized Bound States

A particle is said to be bound when its energy is less than the potential outside the well. Bound-state wavefunctions are normalizable and decay exponentially away from the potential.

---

### Quantum Tunneling

Unlike a classical particle, a quantum particle has a nonzero probability of being found in a region where

$$
E<V(x).
$$

This exponential penetration into forbidden regions is known as **tunneling** and is one of the defining features of quantum mechanics.

---

### Boundary Conditions

Solutions to the Schrödinger equation must satisfy the appropriate boundary conditions.

For the finite square well,

- the wavefunction is continuous,
- its first derivative is continuous.

For the delta-function potential,

- the wavefunction remains continuous,
- its derivative has a finite discontinuity determined by the strength of the delta function.

---

### Numerical Root Finding

The allowed energies of the finite square well satisfy transcendental equations that generally cannot be solved analytically.

Instead, numerical methods such as

```python
scipy.optimize.root_scalar()
```

or

```python
scipy.optimize.brentq()
```

are used to determine the bound-state energies.

## Mathematical Background

The time-independent Schrödinger equation is

$$
-\frac{\hbar^2}{2m}
\frac{d^2\psi}{dx^2}
-
\alpha\delta(x)\psi(x)
=
E\psi(x).
$$

For a bound state,

$$
E<0.
$$

It is convenient to write

$$
E=-\frac{\hbar^2\kappa^2}{2m},
\qquad
\kappa>0.
$$

For $x\neq0$, the potential vanishes, so the Schrödinger equation becomes

$$
-\frac{\hbar^2}{2m}
\frac{d^2\psi}{dx^2}
=
E\psi(x).
$$

Substituting the bound-state energy gives

$$
\frac{d^2\psi}{dx^2}
=
\kappa^2\psi.
$$

Consider the symmetric finite square well

## Mathematical Background

For a bound state,

$$
-V_0<E<0.
$$

Inside the well, define

$$
k
=
\frac{\sqrt{2m(E+V_0)}}{\hbar}.
$$

Outside the well, define

$$
\kappa
=
\frac{\sqrt{-2mE}}{\hbar}.
$$

Because

$$
E+V_0>0,
$$

the wavefunction is oscillatory inside the well.

Because

$$
E<0,
$$

the wavefunction decays exponentially outside the well.

The two wave numbers satisfy

$$
k^2+\kappa^2
=
\frac{2mV_0}{\hbar^2}.
$$

## Numerical Note

A sign change alone does not always imply a physical root if the interval contains a singularity. Before calling the root solver, confirm that

- both endpoints lie on the same continuous branch,
- the function is finite at both endpoints,
- the function changes sign across the interval.

This distinction is essential when solving transcendental equations numerically.
