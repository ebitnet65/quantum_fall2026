# Problem 4 — Dephasing and amplitude damping (25 points)

Keep the same basis and Bloch-vector convention, and use dimensionless time $s=\omega_{\mathrm{ref}}t$. Define dimensionless rates $\gamma_1=\Gamma_1/\omega_{\mathrm{ref}}$ and $\gamma_\phi=\Gamma_\phi/\omega_{\mathrm{ref}}$, where $\Gamma_1$ is the excited-state population decay rate and $\Gamma_\phi$ is the **pure-dephasing contribution to the coherence decay rate**. Let

$$
\gamma_2=\frac{\gamma_1}{2}+\gamma_\phi.
$$

The dissipative Bloch equations are

$$
\frac{dr_x}{ds}=w_y r_z-w_z r_y-\gamma_2r_x,
$$
$$
\frac{dr_y}{ds}=w_z r_x-w_x r_z-\gamma_2r_y,

$$
